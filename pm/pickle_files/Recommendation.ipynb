### 여기부터 피클 차용..

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

data = pd.read_csv('df_for_model.csv')

# keyword 리스트를 문자열로 변환
data['keyword_str'] = data['KeyWords'].apply(lambda x: ' '.join(eval(x)))

# TF-IDF 벡터화
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(data['keyword_str'])
```

```python
# 유사 제품 추천 함수 정의 (가격 필터링x)
def recommend_products_keyword(index, num_recommendations=5):
    # 주어진 제품과 다른 제품들 간의 코사인 유사도 계산
    cosine_sim = cosine_similarity(tfidf_matrix[index], tfidf_matrix).flatten()
    
    # 입력한 제품의 인덱스를 제외하고 유사한 제품의 인덱스 추출
    similar_indices = cosine_sim.argsort()[::-1]  # 유사도 높은 순으로 정렬
    similar_indices = similar_indices[similar_indices != index]  # 자기 자신 제외
    
    # 가장 유사한 제품 5개 추출
    top_similar_indices = similar_indices[:num_recommendations]
    
    # 유사도 값과 함께 추천 제품 반환 (유사도 높은 순서대로)
    recommendations = data.iloc[top_similar_indices][['category1', 'category2', 'category3', 'goodsName', 'KeyWords']].copy()
    recommendations['similarity'] = cosine_sim[top_similar_indices]  # 유사도 값 추가
    return recommendations
```

```python
# 사용자 입력을 받아 제품 추천 (스크롤링 파트)
while True:
    try:
        # 사용자에게 인덱스 입력 받기
        user_input = int(input(f"추천받을 제품의 인덱스 번호를 입력하세요 (0 ~ {len(data)-1}): "))
        
        # 입력 값이 데이터 범위 내에 있는지 확인
        if 0 <= user_input < len(data):
            # 추천 시스템 실행 및 결과 출력
            recommendations = recommend_products_keyword(user_input, num_recommendations=5)
            print("\n유사한 제품 추천 결과:")
            print(recommendations[['category1', 'category2', 'category3', 'goodsName', 'KeyWords', 'similarity']])
            break  # 올바른 입력이 있으면 루프 종료
        else:
            print(f"유효하지 않은 인덱스입니다. 0에서 {len(data)-1} 사이의 값을 입력하세요.")
            
    except ValueError:
        print("올바른 숫자를 입력하세요.")
```

```python
# 사용자 입력을 받아 제품 추천 #추천멘트부분
while True:
    try:
        # 사용자에게 인덱스 입력 받기
        user_input = int(input(f"추천받을 제품의 인덱스 번호를 입력하세요 (0 ~ {len(data)-1}): "))
        
        # 입력 값이 데이터 범위 내에 있는지 확인
        if 0 <= user_input < len(data):
            # 사용자가 조회한 제품명 가져오기
            selected_product_name = data.iloc[user_input]['goodsName']
            
            # 추천 시스템 실행
            recommendations = recommend_products_keyword(user_input, num_recommendations=5)
            
            # 사용자 메시지 출력
            print(f"\n최근 '{selected_product_name}' 상품을 알아보셨군요, 이 상품들은 어떠신가요?\n")
            
            # 추천 결과 출력
            print(recommendations[['category1', 'category2', 'category3', 'goodsName', 'KeyWords', 'similarity']])
            break  # 올바른 입력이 있으면 루프 종료
        else:
            print(f"유효하지 않은 인덱스입니다. 0에서 {len(data)-1} 사이의 값을 입력하세요.")
            
    except ValueError:
        print("올바른 숫자를 입력하세요.")
```

```python
# 유사 제품 추천 함수 정의 (가격 필터링o)
def recommend_products_discounted_price(index, num_recommendations=5):
    # 주어진 제품과 다른 제품들 간의 코사인 유사도 계산
    cosine_sim = cosine_similarity(tfidf_matrix[index], tfidf_matrix).flatten()
    
    # 입력한 제품의 인덱스를 제외하고 유사한 제품의 인덱스 추출
    similar_indices = cosine_sim.argsort()[::-1]  # 유사도 높은 순으로 정렬
    similar_indices = similar_indices[similar_indices != index]  # 자기 자신 제외
    
    # 선택한 제품의 할인된 가격
    selected_product_price = data.iloc[index]['discountedPrice']

    # 유사도가 높은 제품 중에서 할인된 가격이 선택한 제품의 가격보다 낮거나 같은 제품만 필터링
    filtered_indices = [i for i in similar_indices if data.iloc[i]['discountedPrice'] <= selected_product_price]

    # 필터링된 인덱스를 기반으로 유사도가 높은 순서로 상위 N개의 제품 추출
    top_filtered_indices = filtered_indices[:num_recommendations]

    # 유사도 값과 함께 추천 제품 반환 (유사도 높은 순서대로)
    recommendations = data.iloc[top_filtered_indices][['category1', 'category2', 'category3', 'goodsName', 'KeyWords', 'originalPrice',
       'discountedPrice']].copy()
    recommendations['similarity'] = cosine_sim[top_filtered_indices]  # 유사도 값 추가
    return recommendations
```

```python
#가격이 봤던 상품보다 낮은 것으로 필터링해서 추천
while True:
    try:
        # 사용자에게 인덱스 입력 받기
        user_input = int(input(f"추천받을 제품의 인덱스 번호를 입력하세요 (0 ~ {len(data)-1}): "))
        
        # 입력 값이 데이터 범위 내에 있는지 확인
        if 0 <= user_input < len(data):
            # 사용자가 조회한 제품명 가져오기
            selected_product_name = data.iloc[user_input]['goodsName']
            # 사용자가 조회한 제품의 discounted_price 가져오기
            selected_product_discounted_price = data.iloc[user_input]['discountedPrice']
        
            # 추천 시스템 실행
            recommendations = recommend_products_discounted_price(user_input, num_recommendations=5)
            
            # 사용자 메시지 출력
            print(f"\n최근 '{selected_product_name}' 상품을 알아보셨군요, '{selected_product_discounted_price}'원 보다 더 낮은 가격의 상품이 있습니다.\n")
            
            # 추천 결과 출력
            print(recommendations[['category1', 'category2', 'category3', 'goodsName', 'KeyWords', 'originalPrice',
       'discountedPrice', 'similarity']])
            break  # 올바른 입력이 있으면 루프 종료
        else:
            print(f"유효하지 않은 인덱스입니다. 0에서 {len(data)-1} 사이의 값을 입력하세요.")
            
    except ValueError:
        print("올바른 숫자를 입력하세요.")
```
